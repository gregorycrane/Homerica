```python
import re

fname = 'pharr2025.xml'
outfname = '../../Downloads/pharr2025v.xml'

f = open(fname)
outf = open(outfname,'w')

invoc = 0
for l in f:
    l = re.sub('\s+$','',l)
    if(re.search('<list type="gloss">',l)):
        print('<div type="vocab">',file=outf)
        invoc = 1
        continue
    if(re.search('</list>',l) and invoc):
        print('</div>',file=outf)
        invoc = 0
        continue
        

    if(invoc):
        l = re.sub('<item([^>]+>.+)</item>','<div type="newword"\g<1></div>',l)

    print(l,file=outf)
f.close()
outf.close()
```
